```python
import pandas as pd
```

```python
#1° Passo: Importar os dados
df = pd.read_excel('partidas.xlsx')
```

```python
df.head()
```

```python
#2° Passo: Enriquecimento: Criando a coluna de continentes (Dicionários!)
continentes = {
    # América do Sul (CONMEBOL)
    'Argentina': 'South America', 'Brazil': 'South America',
    'Ecuador': 'South America', 'Uruguay': 'South America',

    # Europa (UEFA)
    'Belgium': 'Europe', 'Croatia': 'Europe', 'Denmark': 'Europe',
    'England': 'Europe', 'France': 'Europe', 'Germany': 'Europe',
    'Netherlands': 'Europe', 'Poland': 'Europe', 'Portugal': 'Europe',
    'Serbia': 'Europe', 'Spain': 'Europe', 'Switzerland': 'Europe',
    'Wales': 'Europe',

    # Ásia (AFC)
    'Australia': 'Asia',  'IR Iran': 'Asia', 'Japan': 'Asia',
    'Korea Republic': 'Asia', 'Qatar': 'Asia', 'Saudi Arabia': 'Asia',


    # África (CAF)
    'Cameroon': 'Africa', 'Tunisia': 'Africa','Morocco': 'Africa',
    'Senegal': 'Africa', 'Ghana': 'Africa',

    # América do Norte (CONCAF)
    'Canada': 'North America', 'Mexico': 'North America',
    'United States': 'North America', 'Costa Rica': 'North America',
}

df['home_continent'] = df['home_team'].map(continentes)
df['away_continent'] = df['away_team'].map(continentes)
```

```python
df.head()
```

```python
#3° Passo: Realizar o tratamento dos dados: Extraindo os gols do placar (string manipulation)
#O placar vem como "2-0" ou "(4) 3-3 (2)". Vamos simplificar pegando os números.

def extrair_gols(placar):
  import re
  #Procura por padrão de número - traço - número
  match = re.search(r'(\d+)[\–\-](\d+)', str(placar))
  if match:
      return int(match.group(1)), int(match.group(2))
  return 0, 0

#Aplicando a função
gols = df['score'].apply(extrair_gols)
df['home_goals'] = [g[0] for g in gols]
df['away_goals'] = [g[1] for g in gols]
```

```python
#Outra forma de executar o passo 3° passo
'''
Durante a execução do projeto, eu enfrentei o seguinte problema ao executar o código do 3° passo, a média de gols estava retornando
como NaN(Zero). O caractere de separação no placar do arquivo Excel provavelmente é um "en-dash" (–), que é um traço ligeiramente
mais longo que o hífen (-) comum do teclado. Quando eu escrevi o código o editor de atribui a expressão regular um hífer comum.
Como o código procura um caractere e o arquivo tem outro, ele não encontra nada e retorna 0 (que é o valor padrão da nossa função)
'''
#Para resolver esse problema também podemos usar a seguinte solução:
def extrair_gols(placar):
    import re
    # A função findall procura TODOS os grupos de números na string. Não importa se o separador é traço, espaço ou ponto
    numeros = re.findall(r'\d+', str(placar))

    # Cenário 1: Placar normal (ex: "2-1") -> Encontra ['2', '1']
    if len(numeros) == 2:
        return int(numeros[0]), int(numeros[1])

    #Cenário 2: Placar com pênaltis (ex: "(4) 3-3 (2)") -> Encontra ['4', '3', '3', '2']
    elif len(numeros) == 4:
        return int(numeros[1]), int(numeros[2])
    return 0, 0

gols = df['score'].apply(extrair_gols)
df['home_goals'] = [g[0] for g in gols]
df['away_goals'] = [g[1] for g in gols]

print(df[['score', 'home_goals', 'away_goals']].head())
```

```python
df.head()
```

```python
#4° Passo: Realizar a reestruturação "Empilhando" os dados para comparar continentes
#Vamos criar um DataFrame onde cada linha é um time, não uma partida. Isso facilita muito o cálculo de médias por continentes.

cols_interesse = [
    'continent', 'goals', 'xg', 'possession', 'SoT', 'tackles',
    'clearances', 'long_balls', 'passes_completed', 'saves', 'interceptions',
    'aerials_won', 'fouls'
]

#Dados do time da casa
home_data = df[['home_continent', 'home_goals', 'home_xg', 'home_possession', 'home_sot', 'home_tackles', 'home_clearances', 'home_long_balls', 'home_completed_passes', 'home_saves', 'home_interceptions', 'home_aerials_won', 'home_fouls']].copy()
home_data.columns = cols_interesse

#Dados do time visitante
away_data = df[['away_continent', 'away_goals', 'away_xg', 'away_possession', 'away_sot', 'away_tackles', 'away_clearances', 'away_long_balls', 'away_completed_passes', 'away_saves', 'away_interceptions', 'away_aerials_won', 'away_fouls']].copy()
away_data.columns = cols_interesse

#Cria o dataFrame com todas as informações
df_times = pd.concat([home_data, away_data])
```

```python
df_times.head()
```

```python
#5. Agregação e análise final
analise = df_times.groupby('continent').mean().round(2)

print("--- Comparativo de Médias por Continente (Copa 2022) ---")
print(analise.sort_values('xg', ascending=False))
```

```python
#Confronto direto:
# --- Análise de Confronto Direto: Europa vs América do Sul ---

# 1. Definindo os rivais
rival1 = 'Europe'
rival2 = 'South America'

# 2. Criando o Filtro (Máscara)
# Queremos jogos onde: (Casa é EU e Visitante é SA) OU (Casa é SA e Visitante é EU)
filtro_confronto = (
    ((df['home_continent'] == rival1) & (df['away_continent'] == rival2)) |
    ((df['home_continent'] == rival2) & (df['away_continent'] == rival1))
)

# 3. Aplicando o filtro para criar um DataFrame só desses jogos
confrontos = df[filtro_confronto].copy()

# 4. Calculando quem ganhou cada jogo (Baseado nos gols extraídos)
def definir_vencedor(linha):
    if linha['home_goals'] > linha['away_goals']:
        return linha['home_continent']
    elif linha['away_goals'] > linha['home_goals']:
        return linha['away_continent']
    else:
        return 'Empate'

confrontos['vencedor'] = confrontos.apply(definir_vencedor, axis=1)

# 5. Exibindo os Resultados
print(f"Total de Confrontos Diretos: {len(confrontos)}")
print("\n--- Lista de Jogos ---")
cols_display = ['home_team', 'away_team', 'score', 'vencedor']
print(confrontos[cols_display].to_markdown(index=False))

print("\n--- Placar Geral (Quem ganhou mais?) ---")
placar_geral = confrontos['vencedor'].value_counts()
print(placar_geral)

#Cálculo de Aproveitamento
vitorias_sa = placar_geral.get('South America', 0)
vitorias_eu = placar_geral.get('Europe', 0)
empates = placar_geral.get('Empate', 0)

print(f"\nResultado da análise:")
print(f"América do Sul venceu {vitorias_sa} vezes.")
print(f"Europa venceu {vitorias_eu} vezes.")
print(f"Houveram {empates} empates (incluindo decisões por pênaltis).")
```
